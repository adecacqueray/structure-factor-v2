```python
import structure_factor_v2 as sf
import matplotlib.pyplot as plt

N, D = 10_000, 2
x = sf.korobov_lattice(N, D)           
y = sf.jittered_lattice(N, D)
z = sf.curled_lattice(N, D)

def show_result(x, title=None):
    fig, ax = plt.subplots(1, 2, figsize=(13, 5.5))
    sf.plot(x, auto_zoom=True, max_points=1000, ax=ax[0], return_fig=True)
    ax[0].set_aspect("equal")
    sf.plot_structure_factor(x, ax=ax[1], return_fig=True)
    ax[1].set_xlim(right=2)                       
    ax[1].set_xlabel("$k$"); ax[1].set_ylabel("$S(k)$")
    if title:
        fig.suptitle(title)
    plt.tight_layout()
    plt.show()

show_result(x, "Korobov")
show_result(y, "jittered")
show_result(z, "curled")
```
